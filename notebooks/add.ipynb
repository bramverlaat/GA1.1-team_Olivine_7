```python
def add(a,d):
    return a + d
```

```python
add(3,4)
```

```python
add(1,4)
```

```python
def sum_list(values):
    sum = 0
    for i in range(len(values)):
        sum = add(sum, values[i])
    return sum
```

```python
sum_list([1, 2, 3])
```

```python
sum_list([5, 5, 5, 5])
```
